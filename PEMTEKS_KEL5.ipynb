1. ONE HOT ENCODING

```python
import pandas as pd
from sklearn.preprocessing import OneHotEncoder

documents = [
    "karhutla di Indonesia",
    "terjadi karhulta di wilayah Sumatra dan Kalimantan",
    "Kalimantan dan Sumatra"
]

#memisahkan setiap kata dalam dokumen
words = [doc.split() for doc in documents]

#mengambil semua kata unik
all_words = sorted(set(word for doc in words for word in doc))

#mengubah kata menjadi format kolom
data = []
for doc in words:
    row = {word: 0 for word in all_words}
    for word in doc:
        row[word] = 1
    data.append(row)

df = pd.DataFrame(data)
df
```

```python
# 5. FE result
import numpy as np
def get_document_vector(tokens, model):
    valid_vectors = [model.wv[word] for word in tokens if word in model.wv]
    # if not valid_vectors:
    #   return np.zeros(model.vector_size)
    return np.mean(valid_vectors, axis=0)
X_features = np.array([get_document_vector(doc, model) for doc in sentences])
pd.DataFrame(X_features)
```

## Jaccard

```python
# JACARD
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.metrics import jaccard_score
from scipy.spatial.distance import pdist

# contoh data teks
documents = [
    "karhutla di Indonesia",
    "terjadi karhulta di wilayah Sumatra dan Kalimantan",
    "Kalimantan dan Sumatra"
]

vectorizer = CountVectorizer(binary=True)
X = vectorizer.fit_transform(documents).toarray()

# menghitung jarak Jaccard
distances = pdist(X, metric='jaccard')
print(distances)

# menghitung similarity antar document
Sd1d2 = jaccard_score(X[0], X[1]) #similarity doc 1 and doc 2
Sd1d3 = jaccard_score(X[0], X[2]) #similarity doc 1 and doc 3
Sd2d3 = jaccard_score(X[1], X[2]) #similarity doc 2 and doc 3

Sd1d2, Sd1d3, Sd2d3

# similarity = 1 - distance
```

```python

from sklearn.feature_extraction.text import TfidfVectorizer
from Sastrawi.Stemmer.StemmerFactory import StemmerFactory
from nltk.corpus import stopwords
import nltk

# mengunduh daftar stop words
nltk.download('stopwords')
# membuat stemmer untuk bahasa Indonesia
stemmer = StemmerFactory().create_stemmer()
# mengambil daftar stop words bahasa Indonesia
stop_words = set(stopwords.words("indonesian"))

corpus = []
for summary in data["text"].fillna("").astype(str).to_list():
    tokens = summary.lower().split()
    # melakukan stemming dan menghapus stop words
    tokens = [
        stemmer.stem(token)
        for token in tokens
        if token not in stop_words
    ]
    corpus.append(" ".join(tokens))
# menampilkan hasil prapemrosesan teks pertama
print(corpus[0])
# membuat representasi TF-IDF berdasarkan seluruh kumpulan teks
vectorizer = TfidfVectorizer(
    sublinear_tf=True,
    max_features=5000
)
tfidf_matrix = vectorizer.fit_transform(corpus)
# menampilkan jumlah fitur yang digunakan dalam TF-IDF
print("Jumlah fitur:", vectorizer.get_feature_names_out().shape)
# menampilkan 1000 fitur pertama
vectorizer.get_feature_names_out()[:1000]
```

```python

import pandas as pd
from IPython.display import display, HTML
from sklearn.metrics.pairwise import cosine_similarity

# mengubah dokumen menjadi fitur TF-IDF
documents_tfidf_features = vectorizer.transform(corpus)
# menentukan jumlah dokumen teratas yang ingin ditampilkan
TopK = 3
# menentukan kata kunci pencarian dalam bahasa Indonesia
query = "kebakaran hutan dan lahan di konsesi sawit Kalimantan"
# mengubah query menjadi vektor TF-IDF
query_tfidf_features = vectorizer.transform([query])

print("Fitur TF-IDF untuk query:")
print(query_tfidf_features, "\n")

# menghitung nilai Cosine Similarity antara dokumen dan query
similarities = cosine_similarity(
    documents_tfidf_features,
    query_tfidf_features
)
# memilih indeks dokumen dengan nilai kemiripan tertinggi
top_indices = similarities[:, 0].argsort()[::-1][:TopK]
# menampilkan dokumen yang paling relevan
judul = []
teks = []
nilai_kemiripan = []

for doc_index in top_indices:
    judul.append(f"Dokumen {doc_index + 1}")
    teks.append(data.iloc[doc_index]["text"].replace("\n", " "))
    nilai_kemiripan.append(similarities[doc_index, 0])

# membuat tabel hasil pencarian
hasil = pd.DataFrame({
    "Judul": judul,
    "Teks": teks,
    "Nilai Kemiripan": nilai_kemiripan
})

display(HTML(hasil.to_html(index=False)))
```
